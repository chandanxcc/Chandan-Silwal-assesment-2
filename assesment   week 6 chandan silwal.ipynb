Cell 2: Part A (The Basics)

```python
# A1. Create a dictionary with 3 contacts
contacts = {
    "Alice": "0400 123 456",
    "Bob": "0411 999 888",
    "Charlie": "0422 555 666"
}

# A2. Add a new entry (your name)
contacts["Sam"] = "0433 111 222"

# A3. Update an existing contact’s number
contacts["Bob"] = "0455 000 111"

# A4. Print all contacts in a neat format
print("--- Current Contacts ---")
for name, phone in contacts.items():
    print(f"{name}: {phone}")
```

Cell 3: Part B (Nested Structure & Lookup)

```python
# B1. Redesign contacts with phone and email
contacts = {
    "Alice": {"phone": "0400 123 456", "email": "alice@example.com"},
    "Bob": {"phone": "0455 000 111", "email": "bob@test.com"},
    "Charlie": {"phone": "0422 555 666", "email": "charlie@web.com"},
    "Sam": {"phone": "0433 111 222", "email": "sam@personal.com"}
}

# B2. Print contacts neatly
print("--- Detailed Contact List ---")
for name, info in contacts.items():
    print(f"Name: {name}")
    print(f"  Phone: {info['phone']}")
    print(f"  Email: {info['email']}\n")

# B3. Simple lookup feature
search_name = input("Enter a name to look up: ").strip()
if search_name in contacts:
    details = contacts[search_name]
    print(f"Found {search_name}: {details['phone']}, {details['email']}")
else:
    print("Contact not found.")
```
