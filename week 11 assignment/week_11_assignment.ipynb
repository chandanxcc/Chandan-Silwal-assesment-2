## Module 1 – Testing Specification

### Task
Design a function that converts Celsius to Fahrenheit.

### Description
The function converts temperature values from Celsius to Fahrenheit using the formula:
$$\text{F} = \left(\frac{9}{5}\right)\text{C} + 32$$

It should work for:
* Positive numbers
* Negative numbers
* Decimal values

### Example Outputs
* $0^\circ\text{C} \rightarrow 32^\circ\text{F}$
* $25^\circ\text{C} \rightarrow 77^\circ\text{F}$
* $100^\circ\text{C} \rightarrow 212^\circ\text{F}$

### Edge Cases
* $-40^\circ\text{C} \rightarrow -40^\circ\text{F}$ (special equal point)
* Decimal inputs like $36.6^\circ\text{C}$ should be handled correctly.

### Test Strategy
We validate correctness using assertion tests for known values.

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

print("Module 1 Outputs")
print(celsius_to_fahrenheit(0))
print(celsius_to_fahrenheit(25))
print(celsius_to_fahrenheit(100))
print(celsius_to_fahrenheit(-40))

assert celsius_to_fahrenheit(0) == 32
assert celsius_to_fahrenheit(25) == 77
assert celsius_to_fahrenheit(100) == 212
assert celsius_to_fahrenheit(-40) == -40
```

## Module 2 – Debugging Practice

### Problem
Fix a function that counts vowels in a string.

### Issue Found
The function originally did not correctly handle uppercase letters.

### Debugging Process
1. Added print statements to inspect each character.
2. Identified case-sensitivity issue.
3. Fixed by converting characters to lowercase.

### Fix Explanation
Using `.lower()` ensures both uppercase and lowercase vowels are counted correctly.

```python
def count_vowels(text):
    vowels = "aeiou"
    count = 0

    for letter in text:
        if letter.lower() in vowels:
            count += 1

    return count

print("Module 2 Outputs")
print(count_vowels("Hello"))
print(count_vowels("AEIOU"))
print(count_vowels("Python"))

assert count_vowels("Hello") == 2
assert count_vowels("AEIOU") == 5
assert count_vowels("Python") == 1
assert count_vowels("") == 0
```

## Module 3 – Test Strategy Design

### Task
Design a function that validates and formats phone numbers.

### Requirements
* Accept different formats (spaces, dashes, raw input)
* Extract digits only
* Return format: `(XXX) XXX-XXXX`
* Return error message for invalid inputs

### Valid Cases
* `0412345678`
* `041-234-5678`
* `041 234 5678`

### Invalid Cases
* Less than 10 digits
* More than 10 digits
* Letters or symbols
* Empty input

### Expected Error Output
`"Invalid phone number"`

```python
def format_phone(number):
    digits = ""

    for char in number:
        if char.isdigit():
            digits += char

    if len(digits) != 10:
        return "Invalid phone number"

    return f"({digits[:3]}) {digits[3:6]}-{digits[6:]}"

print("Module 3 Outputs")
print(format_phone("0412345678"))
print(format_phone("041-234-5678"))
print(format_phone("041 234 5678"))

assert format_phone("0412345678") == "(041) 234-5678"
assert format_phone("041-234-5678") == "(041) 234-5678"
assert format_phone("041 234 5678") == "(041) 234-5678"
assert format_phone("123") == "Invalid phone number"
assert format_phone("abcd123") == "Invalid phone number"

print("ALL MODULES PASSED SUCCESSFULLY")
```
