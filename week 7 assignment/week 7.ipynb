Reading and Displaying Data

```
# Define the filename
filename = "my_data.txt" # [cite: 59]

# Open and read the file
try:
    with open(filename, "r") as file:
        content = file.read()

    print("=== MY DATA STORY ===") # [cite: 61]
    print(content) # [cite: 15]

except FileNotFoundError:
    print("Error: The file 'my_data.txt' was not found. Please upload it to your environment.")
```

File Statistics

```
# This code calculates two pieces of information about the file [cite: 15]

with open(filename, "r") as file:
    lines = file.readlines()

# 1. Calculate the number of lines
num_lines = len(lines)

# 2. Calculate the total character count
# We use the 'content' variable from the previous cell
total_chars = len(content)

print(f"--- File Statistics ---")
print(f"Number of lines: {num_lines}")
print(f"Total character count: {total_chars}")
```
